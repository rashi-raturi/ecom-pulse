```python
products = spark.read.table("workspace.ecompulse_bronze.products")
```

```python
from pyspark.sql.functions import col, trim
from pyspark.sql.types import StringType
```

```python
string_cols = [f.name for f in products.schema.fields if isinstance(f.dataType, StringType)]

if string_cols:
    products = products.withColumns({name: trim(col(name)) for name in string_cols})
```

```python
import pyspark.sql.functions as F
```

```python
products = products.withColumn(
    "product_category_name",
    F.lower(F.col("product_category_name"))
)
```

```python
products = products.fillna({
    "product_category_name" : "Unknown",
    "product_name_lenght" : 0,
    "product_description_lenght" : 0,
    "product_photos_qty" : 0
})
```

```python
products.write.mode("overwrite").saveAsTable("workspace.ecompulse_silver.products")
```
