```python
payments = spark.read.table("workspace.ecompulse_bronze.payments")
```

```python
import pyspark.sql.functions as F
```

```python
payments = payments.withColumn(
    "payment_type",
    F.coalesce(F.col("payment_type"), F.lit("unknown"))
)
```

```python
payments = payments.withColumn(
    "payment_type",
    F.lower(F.col("payment_type"))
)
```

```python
payments = payments.withColumn(
    "payment_value",
    F.round(F.col("payment_value"), 2)
)
```

```python
payments.write.mode("overwrite").format("delta").saveAsTable("workspace.ecompulse_silver.payments")
```
